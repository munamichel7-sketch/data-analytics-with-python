1. Write a program that accepts a list from user and print the alternate element of list.

```python
numbers = input("Enter list elements separated by space: ").split()

print("Alternate elements:")

for i in range(0, len(numbers), 2):
    print(numbers[i])
```

Question 2: Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method.

```python
numbers = input("Enter list elements separated by space: ").split()

print("Reversed list:")

for i in range(len(numbers) - 1, -1, -1):
    print(numbers[i])
```

Question 3: Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

```python
numbers = input("Enter numbers separated by space: ").split()

numbers = [int(x) for x in numbers]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number =", largest)
```

Question 4: Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.

```python
numbers = input("Enter list elements separated by space: ").split()

last = numbers[-1]

for i in range(len(numbers) - 1, 0, -1):
    numbers[i] = numbers[i - 1]

numbers[0] = last

print("Rotated list:", numbers)
```

Question 5: Write a program that input a string and ask user to delete a given word from a string.

```python
sentence = input("Enter a string: ")
word = input("Enter the word you want to delete: ")

words = sentence.split()

new_sentence = []

for w in words:
    if w != word:
        new_sentence.append(w)

print("New string:", " ".join(new_sentence))
```

Question 6: Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

```python
date = input("Enter date (mm/dd/yyyy): ")

month, day, year = date.split("/")

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

month_name = months[int(month) - 1]

print(month_name, int(day), ",", year)
```

Question 7: Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses"

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_sentence = ""

    for word in words:
        new_sentence = new_sentence + word[0].upper() + word[1:] + " "

    return new_sentence.strip()


sentence = input("Enter a sentence: ")

result = capitalize_words(sentence)

print("New string:", result)
```

Question 8: Find the sum of each row of matrix of size m x n.

```python
m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

matrix = []

for i in range(m):
    row = []

    for j in range(n):
        value = int(input("Enter element: "))
        row.append(value)

    matrix.append(row)

for i in range(m):
    total = 0

    for j in range(n):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

Question 9: Write a program to add two matrices of size n x m.

```python
m = int(input("Enter number of rows: "))
n = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []

print("Enter elements of first matrix:")

for i in range(m):
    row = []

    for j in range(n):
        value = int(input("Enter element: "))
        row.append(value)

    matrix1.append(row)

print("Enter elements of second matrix:")

for i in range(m):
    row = []

    for j in range(n):
        value = int(input("Enter element: "))
        row.append(value)

    matrix2.append(row)

result = []

for i in range(m):
    row = []

    for j in range(n):
        row.append(matrix1[i][j] + matrix2[i][j])

    result.append(row)

print("Sum of matrices:")

for row in result:
    print(row)
```

Question 10: Write a program to multiply two matrices.

```python
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

if c1 != r2:
    print("Matrix multiplication is not possible.")

else:
    matrix1 = []
    matrix2 = []

    print("Enter elements of first matrix:")

    for i in range(r1):
        row = []

        for j in range(c1):
            value = int(input("Enter element: "))
            row.append(value)

        matrix1.append(row)

    print("Enter elements of second matrix:")

    for i in range(r2):
        row = []

        for j in range(c2):
            value = int(input("Enter element: "))
            row.append(value)

        matrix2.append(row)

    result = []

    for i in range(r1):
        row = []

        for j in range(c2):
            total = 0

            for k in range(c1):
                total = total + matrix1[i][k] * matrix2[k][j]

            row.append(total)

        result.append(row)

    print("Multiplication of matrices:")

    for row in result:
        print(row)
```
