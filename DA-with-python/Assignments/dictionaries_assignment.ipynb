Q4 — Count Frequency of Numbers

numbers = [1,2,2,3,3,3,4,4,4,4]

Create frequency dictionary.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1

print(frequency)
```

Q5 — Word Counter

Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print("Word frequency:", frequency)
```

Q6 — Student Marks & Average

marks = {"Math":80, "English":70, "Physics":90}

Calculate total and average.

```python
marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

total = 0

for mark in marks.values():
    total = total + mark

average = total / len(marks)

print("Total =", total)
print("Average =", average)
```

Q7 — Highest Scoring Student

students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}

Find highest scoring student.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = ""
highest_score = 0

for name, score in students.items():
    if score > highest_score:
        highest_score = score
        highest_student = name

print("Highest scoring student:", highest_student)
print("Score:", highest_score)
```

Q8 — Merge Dictionaries

dict1 = {"a":1,"b":2}

dict2 = {"c":3,"d":4}

Merge and print result.

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged = {}

for key, value in dict1.items():
    merged[key] = value

for key, value in dict2.items():
    merged[key] = value

print(merged)
```

Q9 — Mini Project: Contact Book

Add 3 contacts, search by name, print phone or Contact not found.

```python
contacts = {}

for i in range(3):
    name = input("Enter contact name: ")
    phone = input("Enter phone number: ")
    contacts[name] = phone

search_name = input("Enter name to search: ")

if search_name in contacts:
    print("Phone:", contacts[search_name])
else:
    print("Contact not found")
```
