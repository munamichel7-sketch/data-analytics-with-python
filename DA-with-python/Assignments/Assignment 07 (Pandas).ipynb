1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd

# Create a dictionary of 6 employees
data = {
    "Name": ["Ali", "Ahmed", "Sara", "Usman", "Ayesha", "Hamza"],
    "Age": [22, 25, 23, 28, 24, 26],
    "City": ["Karachi", "Lahore", "Islamabad", "Karachi", "Multan", "Lahore"],
    "Salary": [30000, 45000, 35000, 60000, 40000, 50000]
}


df = pd.DataFrame(data)


print("First 3 Rows:")
print(df.head(3))


print("\nShape:")
print(df.shape)


print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

2) Create a Series of 5 numbers with custom index labels ['a','b','c','d','e']. Print the value at label 'c', the values at positions 1 to 3, and the sum and mean of the Series.

```python
import pandas as pd


s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])


print("Value at label c:", s['c'])


print("\nValues at positions 1 to 3:")
print(s.iloc[1:4])

print("\nSum:", s.sum())


print("Mean:", s.mean())
```

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python

df["Bonus"] = df["Salary"] * 0.10


df["Total"] = df["Salary"] + df["Bonus"]

print("After adding Bonus and Total:")
print(df)


df.rename(columns={"City": "Location"}, inplace=True)


df.drop(columns=["Bonus"], inplace=True)

print("\nAfter renaming City and dropping Bonus:")
print(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import numpy as np

# Create a DataFrame with missing values
df_missing = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Usman", "Ayesha"],
    "Age": [22, np.nan, 25, 28, np.nan],
    "Salary": [30000, 40000, np.nan, 50000, 60000],
    "City": ["Karachi", "Delhi", np.nan, "Lahore", "Karachi"]
})


print("Missing Values:")
print(df_missing.isnull().sum())


df_missing["Age"] = df_missing["Age"].fillna(df_missing["Age"].mean())
df_missing["Salary"] = df_missing["Salary"].fillna(df_missing["Salary"].mean())

print("\nAfter Filling Numeric Missing Values:")
print(df_missing)


df_missing.dropna(inplace=True)

print("\nAfter Dropping Rows with Missing Values:")
print(df_missing)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python

sorted_df = df.sort_values(
    by=["Salary", "Age"],
    ascending=[False, True]
)

print("Sorted DataFrame:")
print(sorted_df)

print("\nTop 3 Highest-Paid Employees:")
print(df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
# Find average, number of employees, and maximum salary for each City
result = df.groupby("Location").agg({
    "Salary": ["mean", "count", "max"]
})

print(result)
```
